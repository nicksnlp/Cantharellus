```python
import json

# Function to modify the .jsonl file
def modify_jsonl(input_file_path, output_file_path):
    with open(input_file_path, 'r') as infile, open(output_file_path, 'w') as outfile:
        # Iterate through each line of the file and track the index for "id"
        for idx, line in enumerate(infile, start=1):
            # Parse the current line as a JSON object
            data = json.loads(line)

            # Create a new dictionary with the desired structure
            new_ordered_data = {
                ##Zhang's data
               """ 'id': idx,
                'lang': "EN",
                'model_input': data["model_input"],  # Direct access
                'model_output_text': data["model_output_text"],  # Direct access
                'hard_labels':list([span['start'], span['end']] for span in data['spans']) """

                ##Nick' data
                'id': data['id'].replace("-_unlabeled", ""),
                'lang': data['lang'],
                'model_input': data["model_input"],
                'model_output_text': data["model_output_text"],
                'hard_labels': data['hard_labels']

            }

            # Write the modified line to the output file
            outfile.write(json.dumps(new_ordered_data) + "\n")

# Example usage
input_file_path = '/content/validation_with_spans.jsonl'#'gemini_qa-pairs_24.11.2024.jsonl'  # Replace with your actual input file path
output_file_path = '/content/validation_with_spans_small.jsonl'#'gemini_qa-pairs_nick_lang_24.11.2024.jsonl'  # Replace with the desired output file path

modify_jsonl(input_file_path, output_file_path)
```
