```python
import json

# File paths
input_file_path = '/content/validation_with_spans.jsonl'
output_file_path = '/content/validation_with_spans_correct_id.jsonl'

# Process the file
with open(input_file_path, 'r') as infile, open(output_file_path, 'w') as outfile:
    for line in infile:
        # Parse JSON line
        record = json.loads(line)

        # Modify the 'id' field if necessary
        if 'id' in record and '_unlabeled' in record['id']:
            record['id'] = record['id'].replace('_unlabeled', '')

        # Write the modified record back to the output file
        outfile.write(json.dumps(record) + '\n')

print(f"Processed file saved as {output_file_path}")
```
