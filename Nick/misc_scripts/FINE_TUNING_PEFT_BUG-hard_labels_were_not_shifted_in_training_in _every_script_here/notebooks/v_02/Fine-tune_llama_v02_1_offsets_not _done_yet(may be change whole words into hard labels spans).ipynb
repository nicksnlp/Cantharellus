```python
#https://chatgpt.com/share/679c9b11-4e58-800b-9fc2-d19a4fe9a220

import torch

def preprocess_data(batch, tokenizer, max_length=512):
    """
    Tokenize sentences and align labels to the tokens, including attention masks, input_ids, and tokens.
    """
    # Tokenize all the sentences in the batch at once
    tokenized_input = tokenizer(batch['text'],
                                truncation=True,
                                padding='max_length',
                                max_length=max_length,
                                return_tensors="pt",
                                return_offsets_mapping=True,
                                add_special_tokens=True
                                )

    # Initialize aligned labels list to store labels for each example in the batch
    aligned_labels = []
    original_labels = []  # Store original sentence-level labels for each sentence
    input_ids = tokenized_input['input_ids']  # Get the input_ids for the batch

    # Iterate over each sentence in the batch
    for i, sentence in enumerate(batch['text']):
        hard_labels = batch['hard_labels'][i]  # Get the hard_labels for this sentence
        tokens = tokenizer.convert_ids_to_tokens(input_ids[i])  # Get the tokens
        attention_mask = tokenized_input['attention_mask'][i]  # Get the attention mask
        offset_mapping = tokenized_input['offset_mapping'][i]  # Get character spans

        ## CREATE LABELING ON A CHARACTER LEVEL
        sentence_labels = []  # Initialize labels

        # Assign labels based on character spans
        for idx, (start, end) in enumerate(offset_mapping):
            if start == end:  # Special tokens like [CLS], [SEP], [PAD]
                sentence_labels.append(-100)
            else:
                # Default label is 0 (not inside hard_labels)
                label = 0
                for (label_start, label_end) in hard_labels:
                    if start >= label_start and end <= label_end:
                        label = 1  # Inside a hard label span
                        break
                sentence_labels.append(label)

        aligned_labels.append(sentence_labels)
        original_labels.append(hard_labels)  # Save the original labels (character spans)

    # Return the processed data including input_ids, attention_mask, aligned labels, tokens, and sentences
    return {
        'original_hard_labels': original_labels,  # Character-level labels before alignment
        'input_ids': input_ids,  # Tokenized input IDs
        'tokens': [tokenizer.convert_ids_to_tokens(ids) for ids in input_ids],  # The tokens
        'labels': torch.tensor(aligned_labels),  # Token-aligned labels
        'attention_mask': tokenized_input['attention_mask'],  # Attention mask
    }

```

```python
import json
import glob
import random

def combine_and_shuffle_jsonl(input_folder, output_file):
    """
    Combines multiple .jsonl files from a folder, shuffles the data, and saves to a new .jsonl file.

    Args:
    - input_folder (str): Folder containing .jsonl files.
    - output_file (str): Path to save the shuffled .jsonl file.
    """
    all_data = []

    # Find all jsonl files in the folder
    jsonl_files = glob.glob(f"{input_folder}/**/*.jsonl", recursive=True)

    # Read and load all jsonl files
    for file in jsonl_files:
        with open(file, 'r', encoding='utf-8') as f:
            for line in f:
                all_data.append(json.loads(line))  # Load each line as JSON

    # Shuffle the combined data
    random.shuffle(all_data)

    # Write shuffled data to a new jsonl file
    with open(output_file, 'w', encoding='utf-8') as f:
        for entry in all_data:
            f.write(json.dumps(entry) + "\n")

    print(f"Combined and shuffled data saved to {output_file}")
```

```python

# Inference: Using the Fine-Tuned Model for Inference
def infer_with_model(input_text):
    # Tokenize input text with offset mapping and special tokens
    inputs = tokenizer(input_text,
                       return_tensors="pt",
                       padding=True,
                       truncation=True,
                       max_length=128,
                       return_offsets_mapping=True,
                       add_special_tokens=True
                       )

    # Move input tensors to model's device
    offset_mapping = inputs.pop("offset_mapping")[0].tolist()  # Extract character spans
    inputs = {k: v.to(model.device) for k, v in inputs.items()}  # Send to correct device

    # Predict token labels
    with torch.no_grad():
        outputs = model(**inputs)
        logits = outputs.logits  # Model outputs
    predicted_labels = torch.argmax(logits, dim=-1)[0].tolist()  # Convert to list

    # Get tokenized words
    tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"][0].tolist())

    # Align tokens with labels and offset mappings
    labeled_tokens = []
    hallucinated_words = []

    for token, label, (start, end) in zip(tokens, predicted_labels, offset_mapping):
        if start == 0 and end == 0:  # Skip special tokens (e.g., [CLS], [SEP])
            continue

        labeled_tokens.append((token, label, (start, end)))  # Add character positions

        if label == 1:
            hallucinated_words.append(input_text[start:end])  # Extract hallucinated word

    return hallucinated_words, labeled_tokens
```

```python
# Example usage of the inference function
question = "Which municipalities does the Italian commune of Ponzone border?"
input_text = " YES, YES, << Ponza\n"
full_text = question+"<@@>"+input_text
hallucinated_words, labeled_tokens = infer_with_model(full_text)

# Print the list of hallucinated words
print("Hallucinated words:")
print(hallucinated_words)
print(full_text)
print(labeled_tokens)
```
