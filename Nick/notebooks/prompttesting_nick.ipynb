```python
! pip install google-generativeai
import configparser
import google.generativeai as genai
import json

from google.colab import userdata
Your_API_Key = userdata.get('GOOGLE_API_KEY')

genai.configure(api_key=Your_API_Key)
model = genai.GenerativeModel(model_name="gemini-1.5-flash")


# Read system prompts from config file
prompts = configparser.ConfigParser()
prompts.read('prompt_for_words.env')

# Set system prompt and detailed prompt

my_prompt = prompts.get('TEMPLATES', 'TOPIC')

response = model.generate_content(my_prompt)


print(response.text)


# To save the response in a json file:
'''with open('example_data.json', 'w', encoding='utf-8') as json_file:
    json.dump(response.text, json_file)'''
#sadly the format is not what we want right now in the json file. Some code should be added to fix this issue.
```
